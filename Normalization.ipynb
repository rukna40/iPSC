```python
import os
import pandas as pd
from scipy.io import mmread

def process_sc_data_csv(matrix_path, features_path, barcodes_dir, output_path):
    print("Checking input files and directories...")

    if not os.path.isfile(matrix_path):
        raise FileNotFoundError(f"Matrix file not found: {matrix_path}")
    if not os.path.isfile(features_path):
        raise FileNotFoundError(f"Features file not found: {features_path}")
    if not os.path.isdir(barcodes_dir):
        raise NotADirectoryError(f"Barcodes directory not found: {barcodes_dir}")

    print("Loading matrix...")
    matrix = mmread(matrix_path).tocsc()
    print(f"Matrix shape: {matrix.shape}")

    print("Loading features...")
    features = pd.read_csv(features_path, sep='\t', header=None)
    gene_names = features[1].values
    print(f"Number of genes (features): {len(gene_names)}")

    print("Loading barcodes from folder...")
    cell_names = sorted(os.listdir(barcodes_dir))
    print(f"Number of barcodes (cells): {len(cell_names)}")

    # Additional checks
    if matrix.shape[0] != len(gene_names):
        raise ValueError(f"Matrix row count ({matrix.shape[0]}) does not match number of genes ({len(gene_names)})")
    if matrix.shape[1] != len(cell_names):
        raise ValueError(f"Matrix column count ({matrix.shape[1]}) does not match number of barcodes ({len(cell_names)})")
    if len(cell_names) != len(set(cell_names)):
        raise ValueError("Barcodes (cell names) are not unique!")
    
    print("Filtering rows with all zeros...")
    non_zero_rows_mask = matrix.getnnz(axis=1) > 0  # Get rows with at least one non-zero value
    matrix = matrix[non_zero_rows_mask, :]          # Filter matrix
    gene_names = gene_names[non_zero_rows_mask]     # Sync gene names
    print(f"Filtered matrix shape: {matrix.shape}")

    print("Creating DataFrame...")
    df = pd.DataFrame.sparse.from_spmatrix(matrix, index=gene_names, columns=cell_names)
    print(f"DataFrame shape: {df.shape}")

    print(f"Saving DataFrame to CSV: {output_path}")
    df.sparse.to_dense().to_csv(output_path)
    print("Done.")

# Example usage:

process_sc_data_csv(
    matrix_path="Data/GSE247758_RAW/GSM7900397_iPSC_matrix.mtx/matrix.mtx",
    features_path="Data/GSE247758_RAW/GSM7900397_iPSC_features.tsv/features.tsv",
    barcodes_dir="Data/GSE247758_RAW/GSM7900397_iPSC_barcodes.tsv",
    output_path="Data/GSE247758_RAW/GSM7900397.csv"
)
```

```python
import os
import gzip
import pandas as pd

folder_path = 'Data\GSE139273\GSE139273'

merged_df = None

for file_name in os.listdir(folder_path):
    if file_name.endswith('.gz'):
        file_path = os.path.join(folder_path, file_name)
        with gzip.open(file_path, 'rt') as f:
            df = pd.read_csv(f, sep='\t', dtype=str)
            df = df[['Name', 'Unique gene reads']]
            df.rename(columns={'Unique gene reads': file_name.replace('.txt.gz','')}, inplace=True)
            if merged_df is None:
                merged_df = df
            else:
                merged_df = pd.merge(merged_df, df, on='Name', how='outer')

merged_df.to_csv(os.path.join(folder_path, 'GSE139273.csv'), index=False)
```

```python
import pandas as pd

def preprocess(file):
    df = pd.read_csv(file)  # use delimiter='\t' for tab-separated files
    # Keep columns from 'Annotation.Divergence' onward
    cols = df.columns.tolist()
    start_idx = cols.index('Annotation.Divergence')
    df = df.iloc[:, start_idx:]
    # Keep only first gene name in Annotation.Divergence
    df['Annotation.Divergence'] = df['Annotation.Divergence'].str.split('|').str[0]
    return df

# Read and preprocess your three datasets
df1 = preprocess('Data/GSE113253_AT.csv')
df2 = preprocess('Data/GSE113253_BM.csv')
df3 = preprocess('Data/GSE113253_RNA_HumanPrimaryStromalCell.csv')

# Merge on 'Annotation.Divergence'
merged_df = df1.merge(df2, on='Annotation.Divergence', how='outer')\
               .merge(df3, on='Annotation.Divergence', how='outer')
merged_df = merged_df.fillna(0)
# Save merged output
merged_df.to_csv('Raw/New/GSE113253.csv', index=False)

print(merged_df)
```
