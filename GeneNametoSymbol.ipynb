```r
library(dplyr)

gprofiler_file <- "Data\\68559_alias.csv"
raw_counts_file <- "Data\\GSE68559.csv"
output_file <- "Raw\\GSE68559.csv"

gprofiler_df <- read.csv(gprofiler_file, stringsAsFactors = FALSE)
raw_counts_df <- read.csv(raw_counts_file, stringsAsFactors = FALSE)

# Convert Gene column to character to match initial_alias type
raw_counts_df$Gene <- as.character(raw_counts_df$Gene)
gprofiler_df$initial_alias <- as.character(gprofiler_df$initial_alias)

gprofiler_unique <- gprofiler_df %>%
  distinct(initial_alias, .keep_all = TRUE)

annotated_counts <- raw_counts_df %>%
  left_join(gprofiler_unique[, c("initial_alias", "name")], by = c("Gene" = "initial_alias")) %>%
  mutate(GeneID = name) %>%
  select(GeneID, everything(), -name)

write.csv(annotated_counts, output_file, row.names = FALSE)
```
