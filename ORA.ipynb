```r
perform_GO_ORA <- function(
  deg_file,
  prefix = "DEG_GO_ORA_",
  logfc_cutoff = 1.2,
  padj_cutoff = 0.05,
  keyType = "SYMBOL",
  ont = "BP",
  n_dot = 15,
  n_bar = 15,
  n_emap = 20,
  n_cnet = 10,
  wordcloud_min_freq = 2
) {
  # Load required libraries
  library(clusterProfiler)
  library(org.Hs.eg.db)
  library(enrichplot)
  library(DOSE)
  library(wordcloud)
  library(RColorBrewer)
  library(GOSemSim)
  library(ggupset)
  library(topGO)
  library(Rgraphviz)

  # Create output directory if doesn't exist
  output_dir <- dirname(prefix)
  if (!dir.exists(output_dir)) {
    dir.create(output_dir, recursive = TRUE)
  }

  # Load DEG CSV
  df <- read.csv(deg_file, header = TRUE)
  
  # Filter significant genes
  sig <- subset(df, padj < padj_cutoff & abs(log2FoldChange) >= logfc_cutoff & !is.na(log2FoldChange))
  
  # Gene vector for enrichment (named vector: logFC, names=gene symbols)
  genes <- sig$log2FoldChange
  names(genes) <- sig$gene
  genes <- sort(genes, decreasing = TRUE)
  
  # Perform GO BP over-representation analysis
  ego_bp <- enrichGO(
    gene = names(genes),
    OrgDb = org.Hs.eg.db,
    keyType = keyType,
    ont = ont,
    pAdjustMethod = "BH",
    pvalueCutoff = 0.05,
    readable = TRUE
  )
  
  # Save enrichment results
  write.csv(as.data.frame(ego_bp), file = paste0(prefix, "ORA_GO_BP_results.csv"), row.names = FALSE)
  
  # Dotplot
  png(paste0(prefix, "ORA_GO_BP_dotplot.png"), width = 8, height = 6, units = "in", res = 300)
  suppressMessages(suppressWarnings(print(dotplot(ego_bp, showCategory = n_dot) + ggtitle("GO BP Over-representation"))))
  dev.off()
  
  # Barplot
  png(paste0(prefix, "ORA_GO_BP_barplot.png"), width = 8, height = 6, units = "in", res = 300)
  suppressMessages(suppressWarnings(print(barplot(ego_bp, showCategory = n_bar) + ggtitle("GO BP Barplot"))))
  dev.off()
  
  # Wordcloud
  png(paste0(prefix, "ORA_wordcloud.png"), width = 10, height = 10, units = "in", res = 300)
  suppressWarnings(
    wordcloud(words = ego_bp$Description, freq = ego_bp$Count,
              min.freq = wordcloud_min_freq, random.order = FALSE,
              colors = brewer.pal(8, "Dark2"))
  )
  dev.off()
  
  # Upsetplot
  png(paste0(prefix, "ORA_upsetplot.png"), width = 8, height = 6, units = "in", res = 300)
  suppressMessages(suppressWarnings(print(upsetplot(ego_bp))))
  dev.off()
  
  # Enrichment map plot (emapplot)
  simMatrix <- pairwise_termsim(ego_bp)
  png(paste0(prefix, "ORA_emapplot.png"), width = 8, height = 6, units = "in", res = 300)
  suppressMessages(suppressWarnings(print(emapplot(simMatrix, showCategory = n_emap))))
  dev.off()
  
  # Cnetplot
  png(paste0(prefix, "ORA_cnetplot.png"), width = 10, height = 8, units = "in", res = 300)
  suppressMessages(suppressWarnings(print(cnetplot(ego_bp, categorySize = "pvalue", foldChange = genes, showCategory = n_cnet))))
  dev.off()
  
  # GO graph plot
  png(paste0(prefix, "ORA_gograph.png"), width = 10, height = 8, units = "in", res = 300)
  suppressMessages(suppressWarnings(plotGOgraph(ego_bp)))
  dev.off()
  
  cat("ORA analysis complete. Results and plots saved.\n")
}
```

# COMPARITIVE ORA

```r
perform_filtered_comparative_ORA <- function(
  deg_files,         # Named vector: dataset names and DEG file paths
  gene_list,         # Character vector: genes to filter DEGs on
  prefix = "filtered_compareCluster_ORA_",
  logfc_cutoff = 1.0,
  padj_cutoff = 0.01,
  keyType = "SYMBOL",
  ont = "BP",
  pvalueCutoff = 0.05,
  showCategory = 15
) {
  library(clusterProfiler)
  library(org.Hs.eg.db)
  library(enrichplot)
  
  gene_lists <- list()
  
  for (dataset_name in names(deg_files)) {
    df <- read.csv(deg_files[dataset_name], header = TRUE)
    sig <- subset(df, padj < padj_cutoff & abs(log2FoldChange) >= logfc_cutoff & !is.na(log2FoldChange))
    # Filter DE genes to only those in gene_list
    filtered_genes <- intersect(sig$gene, gene_list)
    cat("Dataset:", dataset_name, "- Number of filtered genes:", length(filtered_genes), "\n")
    gene_lists[[dataset_name]] <- unique(filtered_genes)
  }
  
  comp_ego <- compareCluster(
    geneCluster = gene_lists,
    fun = "enrichGO",
    OrgDb = org.Hs.eg.db,
    keyType = keyType,
    ont = ont,
    pAdjustMethod = "BH",
    pvalueCutoff = pvalueCutoff,
    readable = TRUE
  )

  write.csv(as.data.frame(comp_ego), file = paste0(prefix, "GO_BP_filtered_compareCluster_results.csv"), row.names = FALSE)
  
  png(paste0(prefix, "GO_BP_filtered_compareCluster_dotplot.png"), width = 10, height = 6, units = "in", res = 300)
  print(dotplot(comp_ego, showCategory = showCategory) + ggtitle("Filtered Comparative GO BP Enrichment"))
  dev.off()
  
  png(paste0(prefix, "GO_BP_filtered_compareCluster_upsetplot.png"), width = 8, height = 6, units = "in", res = 300)
  print(upsetplot(comp_ego))
  dev.off()
  
  cat("Filtered comparative ORA complete, results saved.\n")
  
  return(comp_ego)
}
```

```r
deg_files <- c(
  GSE38265 = "ORA/GSE38265/DEGresults_HUES9_vs_iPSC.csv",
  GSE69626 = "ORA/GSE69626/DEG_OSKM_vs_ESC.csv",
  GSE239446 = "ORA/GSE239446/GSE239446_logfc0_DEG_reprogrammed_vs_NSC.csv"
)

my_gene_list <- c(
  "PPP1R12B","KCNJ13","SMAD2","KCNQ1OT1","RBPMS","GPC6","MBOAT1",
  "ZNF488","LRRC2","SMCO4","ZBED3","TNFSF4","EXPH5","PHACTR1",
  "DNAH9","MAK","PART1","SYT10","GAB1","CCND2","UTRN","ZNF875"
)


perform_filtered_comparative_ORA(
  deg_files = deg_files,
  gene_list = my_gene_list,
  prefix = "ORA/filtered_compareCluster_",
  showCategory = 20
)
```

# FILTERING

```r
filter_genes_from_file <- function(expr_file_path) {
  genes_to_keep_raw <- readLines("ORA/genes_list.txt")
  genes_to_keep <- toupper(gsub("\t", "", trimws(genes_to_keep_raw)))
  genes_to_keep_unique <- unique(genes_to_keep)
  cat("Original duplicated genes:\n")
  print(genes_to_keep_raw[duplicated(genes_to_keep_raw) | duplicated(genes_to_keep_raw, fromLast = TRUE)])
  cat("Duplicated genes after cleaning:\n")
  print(genes_to_keep[duplicated(genes_to_keep) | duplicated(genes_to_keep, fromLast = TRUE)])
  expr <- read.csv(expr_file_path, check.names = FALSE, stringsAsFactors = FALSE)
  expr <- expr[!duplicated(expr[, 7]), ]
  rownames(expr) <- toupper(trimws(expr[, 7]))
  expr <- expr[, -7]
  cat("Total genes in expr:", nrow(expr), "\n")
  cat("Total unique genes in list:", length(genes_to_keep_unique), "\n")
  cat("Matching genes:", sum(rownames(expr) %in% genes_to_keep_unique), "\n")
  filtered_expr <- expr[rownames(expr) %in% genes_to_keep_unique, ]
  not_found <- setdiff(genes_to_keep_unique, rownames(expr))
  cat("Genes not found:\n")
  print(not_found)
  path_parts <- strsplit(expr_file_path, "/")[[1]]
  filename <- tail(path_parts, 1)
  dir_path <- paste(path_parts[-length(path_parts)], collapse = "/")
  filtered_filename <- paste0("filtered_", filename)
  output_file <- file.path(dir_path, filtered_filename)
  write.csv(filtered_expr, output_file)
}
```
